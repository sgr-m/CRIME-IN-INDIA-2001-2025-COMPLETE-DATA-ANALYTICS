# COMMUNITYWISE POPULATION CENSUS 2011

```python
# https://censusindia.gov.in/nada/index.php/catalog/11361
```

```python
comm_wise_population = pd.read_csv('Raw Datasets/Supportive Datasets/COMMUNITYWISE POPULATION - DDW00C-01 MDDS converted.csv')

comm_wise_population.info()
```

### Filteration

```python
# keeping total columns
state_total = comm_wise_population[
    comm_wise_population['total_rural_urban'].str.strip().str.lower() == 'total'
].copy()

state_total = state_total[
    [
        'state_code',
        'area_name',
        'total_persons',
        'total_hindu',
        'total_muslim',
        'total_christian',
        'total_sikh',
        'total_buddhist',
        'total_jain',
        'total_others',
        'no_religion_total'
    ]
]


state_total.info()
```

#### % of population

```python
religion_cols = [
    'total_hindu',
    'total_muslim',
    'total_christian',
    'total_sikh',
    'total_buddhist',
    'total_jain',
    'total_others',
    'no_religion_total'
]

for col in religion_cols:
    state_total[col.replace('total_', '') + '_percent'] = (
        state_total[col] / state_total['total_persons'] * 100
    )

state_total.head()
```

```python
# dropping row count
state_comm_percent = state_total.drop(columns=religion_cols)
```

```python
state_comm_percent.head()
```

### Export

```python
state_comm_percent.to_csv("FilteredData/census_2011_religionwise_population.csv", index=False)
```
