## Working with Dataset: crimes_in_india_2001_2023

```python
# https://www.kaggle.com/datasets/umeshchandra789/crimes-in-india
# loading the dataset
crimes_in_india = pd.read_csv('Raw Datasets/Crime Datasets/crimes_in_india_2001_2023.csv')

# attribute information:
crimes_in_india.info()
```

```python
crimes_in_india.state_ut.unique()
```

```python
crimes_in_india['state_ut'] = (
    crimes_in_india['state_ut']
    .str.upper()
    .str.replace('&', 'AND', regex=False)
    .str.replace(r'\s+', ' ', regex=True)
    .str.strip()
)
```

```python
# applying precise mapping
state_mapping = {
    'A AND N ISLANDS': 'ANDAMAN AND NICOBAR ISLANDS',
    'D AND N HAVELI': 'DADRA AND NAGAR HAVELI AND DAMAN AND DIU',
    'DAMAN AND DIU': 'DADRA AND NAGAR HAVELI AND DAMAN AND DIU',
    'DELHI UT': 'DELHI',
    'JAMMU AND KASHMIR': 'JAMMU AND KASHMIR'
}

crimes_in_india['state_ut'] = crimes_in_india['state_ut'].replace(state_mapping)
```

```python
sorted(crimes_in_india['state_ut'].unique())
```

```python
# printing random 20 records
for i in random.randint(0,500, size = 20):
    print(crimes_in_india.state_ut.loc[i])
```

```python
# A. Standardize Identifiers
crimes_in_india['state_ut'] = crimes_in_india['state_ut'].str.upper().str.strip()
crimes_in_india['destrict'] = crimes_in_india['district'].str.upper().str.strip()
```

```python
# B. Sort for Time-Series Consistency
crimes_in_india = crimes_in_india.sort_values(['state_ut', 'district', 'year'])
```

### Violent Crimes

```python
violent_crimes = [
    'murder',
    'attempt_to_murder',
    'culpable_homicide_not_amounting_to_murder',
    'rape',
    'riots',
    'hurt_grievous_hurt',
    'dowry_deaths',
    'causing_death_by_negligence'
]
```

### Crimes Against Women

```python
women_crimes = [
    'rape',
    'custodial_rape',
    'other_rape',
    'kidnapping_and_abduction_of_women_and_girls',
    'dowry_deaths',
    'assault_on_women_with_intent_to_outrage_her_modesty',
    'insult_to_modesty_of_women',
    'cruelty_by_husband_or_his_relatives',
    'importation_of_girls_from_foreign_countries'
]
```

### Property Crimes

```python
property_crimes = [
    'dacoity',
    'preparation_and_assembly_for_dacoity',
    'robbery',
    'burglary',
    'theft',
    'auto_theft',
    'other_theft'
]
```

### Economic / Fraud Crimes

```python
economic_crimes = [
    'criminal_breach_of_trust',
    'cheating',
    'counterfeiting'
]
```

## Adding Derived Attributes to the DataFrame

```python
crimes_in_india['voilent_crime_total'] = crimes_in_india[violent_crimes].sum(axis=1)
crimes_in_india['women_crime_total'] = crimes_in_india[women_crimes].sum(axis=1)
crimes_in_india['property_crime_total'] = crimes_in_india[property_crimes].sum(axis=1)
crimes_in_india['economic_crime_total'] = crimes_in_india[economic_crimes].sum(axis=1)
```

```python
# printing updated dataframe information
crimes_in_india.info()
```

```python
# Exporting the dataframe in New CSV File
crimes_in_india.to_csv('FilteredData/crimes_in_india_2001_2023_filtered.csv')
```

## Working with Dataset : crime_dataset_india_2020-2024.csv

```python
# https://www.kaggle.com/datasets/sudhanvahg/indian-crimes-dataset

# Loading the dataset:
crime_df_2024 = pd.read_csv("Raw Datasets/Crime Datasets/crime_dataset_india_2020-2024.csv")

crime_df_2024.info()
```

```python
# printing first 5 rows
crime_df_2024.head()
```

### Cleaning The Dataset 

```python
# renaming the columns for clean code
crime_df_2024.columns = crime_df_2024.columns.str.lower().str.replace(' ', '_')

#printing first 2 records
crime_df_2024.head(2)
```

```python
date_cols = ['date_reported', 'date_of_occurrence', 'date_case_closed']
for col in date_cols:
    crime_df_2024[col] = pd.to_datetime(crime_df_2024[col], errors='coerce')

crime_df_2024.info()
```

```python
# checking the date time format
crime_df_2024.date_reported.head(10)
```

```python
# checking for NaN values:
crime_df_2024.isna().sum()
```

```python
# filling NaN value for attribute: weapon_used -> UNKNOWN
crime_df_2024['weapon_used'] = crime_df_2024['weapon_used'].fillna('UNKNOWN')
```

## DERIVED COLUMNS

### A. Case Status (Binary)

```python
crime_df_2024['case_closed_flag'] = crime_df_2024['case_closed'].map({
    'Yes': 1,
    'No': 0
})

# Yes matlab 1 and No matlab 0 simple
```

### B. Case Resolution Time

```python
crime_df_2024['case_resolution_days'] = (
    crime_df_2024['date_case_closed'] - crime_df_2024['date_reported']
).dt.days

crime_df_2024.info()
```

```python
# total NaN value in case_resolution_days
crime_df_2024.case_resolution_days.isna().sum()

# here NaN reflect because case is not closed!
# # # # baad me dekhunga...
```

### C. Victim Age Grouping

```python
crime_df_2024['victim_age_group'] = pd.cut(
    crime_df_2024['victim_age'],
    bins=[0, 17, 30, 45, 60, 100],
    labels=['Minor', 'Young Adult', 'Adult', 'Middle Aged', 'Senior']
)

# | Age Range | Group       |
# | --------- | ----------- |
# | 0 – 17    | Minor       |
# | 18 – 30   | Young Adult |
# | 31 – 45   | Adult       |
# | 46 – 60   | Middle Aged |
# | 61 – 100  | Senior      |

crime_df_2024.info()
```

## Working with Dataset: NCRB_Table_1A.1 

#### State/UT-wise Number of Indian Penal Code (IPC) Crimes from 2020 to 2022 - NCRB_Table_1A.1 

```python
# https://www.data.gov.in/resource/stateut-wise-number-indian-penal-code-ipc-crimes-2020-2022

# Loading the dataset:
ncrb_table_ipc_crimes = pd.read_csv('Raw Datasets/Crime Datasets/NCRB_Table_1A.1.csv')

# checking the attributes
ncrb_table_ipc_crimes.info()
```

### Cleaning the Dataset

```python
ncrb = ncrb_table_ipc_crimes.copy()

ncrb.columns = [
    'sl_no',
    'state_ut',
    'crimes_2020',
    'crimes_2021',
    'crimes_2022',
    'population_lakhs_2022',
    'crime_rate_2022',
    'chargesheeting_rate_2022'
]

# transform strings:
ncrb['state_ut'] = ncrb['state_ut'].str.upper().str.strip()
```

### A. Year-on-Year Growth (2020–2022)

```python
# [(cy - py) / py]x 100

ncrb['growth_20_21_pct'] = (
    (ncrb['crimes_2021'] - ncrb['crimes_2020']) / ncrb['crimes_2020']
) * 100

ncrb['growth_21_22_pct'] = (
    (ncrb['crimes_2022'] - ncrb['crimes_2021']) / ncrb['crimes_2021']
) * 100
```

### B. Average Crime Volume

```python
ncrb['avg_crimes_2020_22'] = ncrb[['crimes_2020','crimes_2021','crimes_2022']].mean(axis=1)
```

### C. Crimes per Million Population

```python
# lakh -> million (1m = 10 Lakh)
ncrb['population_million_2022'] = ncrb['population_lakhs_2022'] / 10


# per million = total crimes / population
ncrb['crime_per_million_2022'] = (
    ncrb['crimes_2022'] / ncrb['population_million_2022']
)
```

## Working with Crime Judgement Analysis 

## Loading Data: 

```python
# https://njdg.ecourts.gov.in/njdg_v3/

judge_count = pd.read_excel("Raw Datasets/Crime Datasets/court_judge_count_report 2025.xlsx")

judge_count.info()
```

| Column                          | Meaning                         |
| ------------------------------- | ------------------------------- |
| Presiding Officers / Judges (A) | Actual working judges           |
| Incharge Courts (B)             | Courts without permanent judges |
| Link Courts (C)                 | Courts managed temporarily      |
| Total Courts (A+B+C)            | Total sanctioned courts         |
| Empty Courts                    | Vacant courts                   |
| Invalid JOCode Count            | Data quality / admin issues     |
| Courts Closed but cases pending | Backlog indicator               |

### A. Rename Columns for readability

```python
judges = judge_count.copy()

judges.columns = [
    'sr_no',
    'state_ut',
    'working_judges',
    'incharge_courts',
    'link_courts',
    'total_courts',
    'empty_courts',
    'invalid_jocode_count',
    'courts_closed_cases_pending'
]
```

### B. Cleaning State Names

```python
judges['state_ut'] = judges['state_ut'].str.upper().str.strip()
```

```python
judges.info()
```

```python
# judges['pending_cases_in_closed_courts'] = (
#     judges['courts_closed_cases_pending']
#     .str.extract(r'(\d+)')
#     .astype(float)
# )
```

```python
judges.head()
```

## Derived Metric

### A. Court Vacancy Rate

```python
# Indicate justice delays.
judges['court_vacancy_rate_pct'] = (
    judges['empty_courts'] / judges['total_courts']
) * 100
```

### B. Judges per Court

```python
judges['judges_per_court'] = (
    judges['working_judges'] / judges['total_courts']
)
```

### C. Temporary Dependency Ratio

```python
# TCR = [n(inchageCourts) + n(linkCourts)]/n(totalCourts)
judges['temporary_court_ratio_pct'] = (
    (judges['incharge_courts'] + judges['link_courts']) / judges['total_courts']
) * 100
```

### Loading dataset: Summary_Report_on_Courts_2025

```python
# https://njdg.ecourts.gov.in/njdg_v3/ 
summary_report_on_courts = pd.read_excel("Raw Datasets/Crime Datasets/Summary_Report_on_Courts_2025.xlsx")

summary_report_on_courts.info()
```

| Column                | Column Information               |
| --------------------- | -------------------------------- |
| Total Districts       | Administrative spread            |
| Total Court Complexes | Physical infrastructure          |
| Total Establishments  | Functional courts                |
| Total Cases in DB     | Case burden (pending + disposed) |
| Total Orders          | Judicial output                  |
| Orders Not Uploaded   | Digital inefficiency             |


## Cleaning & Standardization

### A. Rename Columns

```python
courts = summary_report_on_courts.copy()

courts.columns = [
    'sr_no',
    'state_ut',
    'total_districts',
    'court_complexes',
    'total_establishments',
    'total_cases_db',
    'total_orders',
    'orders_not_uploaded'
]
```

### B. Normalize State Names

```python
courts['state_ut'] = courts['state_ut'].str.upper().str.strip()
```

## High-Impact Derived Metrics

### A. Case Load per Establishment

```python
# pressure per court = total case / total establishment

courts['cases_per_establishment'] = (
    courts['total_cases_db'] / courts['total_establishments']
)
```

### B. Orders per Establishment

```python
# productivity = total orders / establishment

courts['orders_per_establishment'] = (
    courts['total_orders'] / courts['total_establishments']
)
```

### C. Digital Efficiency Gap (% Orders Not Uploaded)

```python
# useful for e-governance analysis

courts['orders_not_uploaded_pct'] = (
    courts['orders_not_uploaded'] / courts['total_orders']
) * 100
```

### D. Cases per District (Regional Load)

```python
# total cases / no of disticts

courts['cases_per_district'] = (
    courts['total_cases_db'] / courts['total_districts']
)
```

## Merge with Judge Infrastructure

```python
justice_capacity = judges.merge(
    courts,
    on='state_ut',
    how='inner'
)
justice_capacity.info()
```

```python
# export 
justice_capacity.to_csv('FilteredData/justice_capacity_in_India_2025.csv')
```
